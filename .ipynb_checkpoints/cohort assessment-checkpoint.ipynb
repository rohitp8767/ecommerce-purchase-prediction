```python
####################  Data Analysis & Calculation  #####################3
import numpy as np   
import pandas as pd  
import datetime  

####################  Visuvalization & plotting  #####################3
import matplotlib.pyplot as plt 
import seaborn as sns
%matplotlib inline  
#################### Machine Learning #####################3
from sklearn.ensemble import AdaBoostClassifier  
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.metrics import confusion_matrix, classification_report
import warnings  
warnings.filterwarnings("ignore")
from sklearn.preprocessing import LabelEncoder 
```

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Product hierarchy
products = {
    'Electronics': {
        'Mobile': ['Android', 'iOS'],
        'Laptop': ['Gaming', 'Ultrabook']
    },
    'Fashion': {
        'Men Shirt': ['Slim', 'Regular'],
        'Jeans': ['Denim', 'Slim'],
        'T-Shirt': ['Round Neck', 'V-Neck']
    },
    'Books': {
        'Novel': ['Fiction', 'Mystery'],
        'Textbook': ['Math', 'Science']
    }
}

# Simulation config
num_rows = 10000
start_time = datetime.now()

# Helper functions
def get_random_category():
    return random.choice(list(products.keys()))

def get_random_product_type(category):
    product = random.choice(list(products[category].keys()))
    prod_type = random.choice(products[category][product])
    return product, prod_type

def simulate_price(category):
    if category == 'Electronics':
        return np.random.uniform(10000, 200000)
    elif category == 'Fashion':
        return np.random.uniform(500, 5000)
    elif category == 'Books':
        return np.random.uniform(200, 1500)

def simulate_clicks():
    return np.random.poisson(5)

def simulate_binary(prob_yes=0.5):
    return 'Yes' if random.random() < prob_yes else 'No'

# Simulate data
data = []
for i in range(num_rows):
    timestamp = start_time + timedelta(seconds=i)
    category = get_random_category()
    product, prod_type = get_random_product_type(category)
    price = round(simulate_price(category), 2)
    num_clicks = simulate_clicks()
    added_to_cart = simulate_binary(0.6)  # Assume 60% chance
    purchased = simulate_binary(0.3 if added_to_cart == 'Yes' else 0.05)  # Higher chance if added

    data.append({
        'timestamp': timestamp,
        'product_category': category,
        'product': product,
        'type': prod_type,
        'price': price,
        'num_clicks': num_clicks,
        'added_to_cart': added_to_cart,
        'purchased': purchased
    })

# Create DataFrame
df = pd.DataFrame(data)

# Show first few rows


# Save to CSV if needed
# df.to_csv('simulated_ecommerce_data.csv', index=False)
```

# Random Forest model 

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier # y = Category
```

### grid search 

```python
params = { 'min_samples_split': [50,100], ## Parent 
           'min_samples_leaf': [25,50], ## Child or Leave 
           'max_depth':[3,4,5]}  # Branches
```

```python
grid_search_cv = GridSearchCV(RandomForestClassifier(random_state=42,class_weight='balanced'),
                              params, 
                              verbose=1, 
                              cv=10)
grid_search_cv.fit(xtrain, ytrain)
```

```python
grid_search_cv.best_estimator_
```

```python
from sklearn.ensemble import RandomForestRegressor  # y = Number
from sklearn.ensemble import RandomForestClassifier # y = Category

# by default 100 tree will be form 
RFModel = RandomForestClassifier(**grid_search_cv.best_estimator_.get_params()
                            # random_state=42,
                            #    n_estimators=25, # make 25 tress
                            #    criterion="gini",
                            #    max_depth=3,  # each tree will have 4 branches
                            #    min_samples_split=50,# each tree will have parent node
                            #    min_samples_leaf=25,# each tree will have Child node
                            #    max_features="sqrt"
                                )# Variable selection in each tree  

RFModel.fit(xtrain, ytrain)
```

```python
train=pd.concat([xtrain,ytrain],axis=1)
```

# Ada Boost

### grid search method 

```python
model_parameters = {'n_estimators': [30, 50, 100, 150],  # number of Tree (Weak Model)
                    'learning_rate': [ 0.4,0.6, 0.8, 1]}
```

```python
abc = AdaBoostClassifier() 
gscv_ADA = GridSearchCV(estimator=abc, 
                    param_grid=model_parameters, 
                    cv=10, 
                    verbose=1, ### message how many task done at backend
                    n_jobs=-1, ## parallel processing used entire processor
                    scoring='accuracy') ## accuracy evaluation

gscv_ADA.fit(xtrain,ytrain)
```

```python
print('The best parameter are -', gscv_ADA.best_params_)
```

```python
# Re-fit the model with the best parameters
final_mod_ADA = AdaBoostClassifier(**gscv_ADA.best_params_)
final_mod_ADA.fit(xtrain,ytrain)
```

# prediction train

```python
train_pred = final_mod_ADA.predict(xtrain)
```
